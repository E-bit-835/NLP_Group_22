# Notebook 01 - Load All Datasets

## Objective
This notebook loads all datasets used in the Mobixa NLP Chatbot project and performs an initial inspection of each dataset before preprocessing.

```python
import pandas as pd
import numpy as np

print("Libraries imported successfully.")
```

```python
products = pd.read_csv("../data/Product Dataset.csv")

accessories = pd.read_csv("../data/mobile_accessories.csv")

faq = pd.read_csv("../data/mobixa_faq.csv")

intents = pd.read_csv("../data/mobixa_intents.csv")

repairs = pd.read_csv("../data/repair_services.csv")

entities = pd.read_csv("../data/entities.csv")

print("All datasets loaded successfully.")
```
