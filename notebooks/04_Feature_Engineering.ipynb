# Notebook 04 - Feature Engineering

This notebook prepares text data for Machine Learning and Deep Learning models.

Techniques:
- Tokenization
- TF-IDF
- Bag of Words
- Label Encoding
- Train/Test Split

```python
import pandas as pd

from sklearn.feature_extraction.text import CountVectorizer
from sklearn.feature_extraction.text import TfidfVectorizer

from sklearn.preprocessing import LabelEncoder

from sklearn.model_selection import train_test_split
```

```python
intents = pd.read_csv("../dataset/intents_clean.csv")
```

```python
intents.head()
```

```python
intents.columns
```

```python
X = intents["text"]

y = intents["intent"]
```

```python
bow = CountVectorizer()
```

```python
X_bow = bow.fit_transform(X)
```

```python
print(X_bow.shape)
```

```python
bow_df = pd.DataFrame(
    X_bow.toarray(),
    columns=bow.get_feature_names_out()
)

bow_df.head()
```

```python
tfidf = TfidfVectorizer()
```

```python
X_tfidf = tfidf.fit_transform(X)
```

```python
print(X_tfidf.shape)
```

```python
tfidf_df = pd.DataFrame(
    X_tfidf.toarray(),
    columns=tfidf.get_feature_names_out()
)

tfidf_df.head()
```

```python
encoder = LabelEncoder()
```

```python
y_encoded = encoder.fit_transform(y)
```

```python
print(y_encoded)
```

```python
mapping = pd.DataFrame({

"Intent":encoder.classes_,

"Encoded Value":range(len(encoder.classes_))

})

mapping
```

```python
X_train,X_test,y_train,y_test=train_test_split(

X_tfidf,

y_encoded,

test_size=0.2,

random_state=42
)
```

```python
print(X_train.shape)

print(X_test.shape)
```

```python
print(len(y_train))

print(len(y_test))
```
