# Notebook 02 - Data Preprocessing

This notebook performs data cleaning for all Mobixa chatbot datasets.

Datasets:
- Product Dataset
- Mobile Accessories
- Repair Services
- FAQ
- Intent Dataset
- NER Dataset

```python
import pandas as pd
import numpy as np
import re
import string

import nltk

from nltk.corpus import stopwords
from nltk.tokenize import word_tokenize
from nltk.stem import WordNetLemmatizer
```

```python
nltk.download("punkt")
nltk.download("stopwords")
nltk.download("wordnet")
```

```python
products = pd.read_csv("../dataset/Product Dataset.csv")

accessories = pd.read_csv("../dataset/mobile_accessories.csv")

repairs = pd.read_csv("../dataset/repair_services.csv")

faq = pd.read_csv("../dataset/mobixa_faq.csv")

intents = pd.read_csv("../dataset/mobixa_intents.csv")

entities = pd.read_csv("../dataset/entities.csv")
```

```python
stop_words = set(stopwords.words("english"))

lemmatizer = WordNetLemmatizer()
```

```python
def remove_spaces(df):

    return df.apply(
        lambda col: col.str.strip()
        if col.dtype=="object"
        else col
    )


def lowercase(df):

    return df.apply(
        lambda col: col.str.lower()
        if col.dtype=="object"
        else col
    )
```

```python
def clean_text(text):

    text = text.lower()

    text = re.sub(r"http\S+|www\S+","",text)

    text = text.translate(
        str.maketrans("","",string.punctuation)
    )

    words = word_tokenize(text)

    words = [
        word for word in words
        if word not in stop_words
    ]

    words = [
        lemmatizer.lemmatize(word)
        for word in words
    ]

    return " ".join(words)
```
