# Notebook 03 - Exploratory Data Analysis (EDA)

This notebook analyzes cleaned Mobixa chatbot datasets.

Datasets:
- Product Dataset
- Mobile Accessories Dataset
- Repair Services Dataset
- FAQ Dataset
- Intent Dataset
- NER Dataset

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
```

```python
products = pd.read_csv("../data/products_clean.csv")

accessories = pd.read_csv("../data/accessories_clean.csv")

repairs = pd.read_csv("../data/repair_services_clean.csv")

faq = pd.read_csv("../data/faq_clean.csv")

intents = pd.read_csv("../data/intents_clean.csv")

entities = pd.read_csv("../data/entities_clean.csv")
```

```python
print("Products:", products.shape)

print("Accessories:", accessories.shape)

print("Repairs:", repairs.shape)

print("FAQ:", faq.shape)

print("Intents:", intents.shape)

print("Entities:", entities.shape)
```

```python
products.columns
```

```python
products.head()
```

```python
products["Model Name"].value_counts()
```

```python
import matplotlib.pyplot as plt

model_counts = products["Model Name"].value_counts().head(20)

plt.figure(figsize=(10,8))

model_counts.sort_values().plot(
    kind="barh"
)

plt.title("Top 20 Mobile Phone Models")
plt.xlabel("Number of Products")
plt.ylabel("Model Name")

plt.tight_layout()
plt.show()
```

```python
products["Company Name"].value_counts().head(10)
```

```python
plt.figure(figsize=(10,5))

products["Company Name"].value_counts().head(10).plot(
    kind="bar"
)

plt.title("Top 10 Mobile Brands")

plt.xlabel("Brand")

plt.ylabel("Count")

plt.xticks(rotation=45)

plt.show()
```

```python
accessories.columns
```

```python
accessories["category"].value_counts()
```

```python
plt.figure(figsize=(10,5))

accessories["category"].value_counts().plot(
    kind="bar"
)

plt.title("Accessory Categories")

plt.xlabel("Category")

plt.ylabel("Count")

plt.xticks(rotation=45)

plt.show()
```

```python
repairs["repair_service"].value_counts()
```

```python
plt.figure(figsize=(14,6))

repairs["repair_service"].value_counts().plot(
    kind="bar"
)

plt.title("Repair Service Types")

plt.xlabel("Service")

plt.ylabel("Count")

plt.xticks(rotation=70)

plt.show()
```

```python
faq.head()
```

```python
print("Number of FAQ Questions:",
      len(faq))
```

```python
faq["Question_Length"] = faq["question"].apply(
    lambda x: len(str(x).split())
)
```

```python
faq["Question_Length"].describe()
```

```python
plt.figure(figsize=(8,5))

faq["Question_Length"].plot(
    kind="hist",
    bins=20
)

plt.title("FAQ Question Length")

plt.xlabel("Words")

plt.ylabel("Frequency")

plt.show()
```

```python
intents.head()
```

```python
intents["intent"].value_counts()
```

```python
plt.figure(figsize=(12,5))

intents["intent"].value_counts().plot(
    kind="bar"
)

plt.title("Intent Distribution")

plt.xlabel("Intent")

plt.ylabel("Count")

plt.xticks(rotation=45)

plt.show()
```

```python
entities["entity_type"].value_counts()
```

```python
plt.figure(figsize=(8,5))

entities["entity_type"].value_counts().plot(
    kind="bar"
)

plt.title("NER Entity Distribution")

plt.xlabel("Entity Type")

plt.ylabel("Count")

plt.xticks(rotation=45)

plt.show()
```

```python
summary = pd.DataFrame({

"Dataset":[
"Products",
"Accessories",
"Repairs",
"FAQ",
"Intent",
"NER"
],

"Rows":[
len(products),
len(accessories),
len(repairs),
len(faq),
len(intents),
len(entities)
]

})

summary
```
