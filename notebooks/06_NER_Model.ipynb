```python
import pandas as pd
import re
import ast
```

```python
entities = pd.read_csv("../data/entities.csv")

entities.head()
```

```python
# Remove columns that start with "Unnamed"

entities = entities.loc[:, ~entities.columns.str.contains("^Unnamed")]

entities.head()
```

```python
entities["entity_type"].unique()
```

```python
entity_dictionary = {}

for index, row in entities.iterrows():

    entity_type = row["entity_type"]

    entity_value = row["entity_value"]

    synonyms = row["synonyms"]


    words = []

    words.append(entity_value.lower())


    if pd.notna(synonyms):

        synonyms_list = synonyms.split(",")

        for word in synonyms_list:
            words.append(word.strip().lower())


    if entity_type not in entity_dictionary:
        entity_dictionary[entity_type] = []


    entity_dictionary[entity_type].extend(words)


entity_dictionary
```

```python
def extract_entities(text):

    text = text.lower()

    detected_entities = {}


    for entity_type, words in entity_dictionary.items():

        found = []


        for word in words:

            if re.search(r"\b" + word + r"\b", text):

                found.append(word)


        if found:

            detected_entities[entity_type] = found


    return detected_entities
```

```python
sentence = "I need a Samsung Galaxy charger"

result = extract_entities(sentence)


print(result)
```

```python
examples = [

    "What is the price of iPhone 15?",

    "Do you have Samsung charger?",

    "I need a phone case",

    "Repair my Xiaomi phone"

]


for text in examples:

    print("\nText:",text)

    print("Entities:",extract_entities(text))
```

```python
import pickle


with open("../models/entity_dictionary.pkl","wb") as file:

    pickle.dump(entity_dictionary,file)


print("NER dictionary saved")
```
