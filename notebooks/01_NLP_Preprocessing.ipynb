```python
import pandas as pd
import nltk
import string

from nltk.tokenize import word_tokenize
from nltk.corpus import stopwords
from nltk.stem import WordNetLemmatizer
```

```python
nltk.download("punkt")
nltk.download("stopwords")
nltk.download("wordnet")
```

```python
df = pd.read_csv("Phones_Dataset.csv")
```

```python
df.head()
```

```python
df.columns
```

```python
df["text"] = (
    df["Company Name"].astype(str) + " " +
    df["Model Name"].astype(str) + " " +
    df["Warranty"].astype(str) + " " +
    df["Stock Status"].astype(str)
)
```

```python
df["text"].head()
```

```python
stop_words = set(stopwords.words("english"))

lemmatizer = WordNetLemmatizer()
```

```python
def preprocess_text(text):

    # Convert to lowercase
    text = text.lower()

    # Remove punctuation
    text = text.translate(str.maketrans('', '', string.punctuation))

    # Tokenization
    tokens = word_tokenize(text)

    # Remove stop words
    tokens = [word for word in tokens if word not in stop_words]

    # Lemmatization
    tokens = [lemmatizer.lemmatize(word) for word in tokens]

    # Join back into a sentence
    return " ".join(tokens)
```

```python
sentence = "How much is the Apple iPhone 16?"

print(preprocess_text(sentence))
```

```python
df["processed_text"] = df["text"].apply(preprocess_text)
```

```python
df[["text", "processed_text"]].head()
```

```python
df.to_csv("Phones_Dataset.csv", index=False)
```

```python
import string
from nltk.tokenize import word_tokenize
from nltk.corpus import stopwords
from nltk.stem import PorterStemmer, WordNetLemmatizer

# Load dataset
faq = pd.read_csv("mobixa_faq.csv")

# Convert to lowercase
faq["lowercase"] = faq["question"].str.lower()

# Remove punctuation
faq["no_punctuation"] = faq["lowercase"].apply(
    lambda x: x.translate(str.maketrans('', '', string.punctuation))
)

# Tokenization
faq["tokens"] = faq["no_punctuation"].apply(word_tokenize)

# Remove stop words
stop_words = set(stopwords.words("english"))

faq["no_stopwords"] = faq["tokens"].apply(
    lambda words: [word for word in words if word not in stop_words]
)

# Stemming
stemmer = PorterStemmer()

faq["stemmed"] = faq["no_stopwords"].apply(
    lambda words: [stemmer.stem(word) for word in words]
)

# Lemmatization
lemmatizer = WordNetLemmatizer()

faq["lemmatized"] = faq["no_stopwords"].apply(
    lambda words: [lemmatizer.lemmatize(word) for word in words]
)

# Final processed text
faq["processed_text"] = faq["lemmatized"].apply(
    lambda words: " ".join(words)
)

# Save cleaned dataset
faq.to_csv("mobixa_faq_clean.csv", index=False)

# View results
print(faq[["question", "processed_text"]].head())
```
