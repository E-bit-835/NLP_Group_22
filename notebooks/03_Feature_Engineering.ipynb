```python
import pandas as pd
import pickle

from sklearn.feature_extraction.text import CountVectorizer
from sklearn.feature_extraction.text import TfidfVectorizer

from tensorflow.keras.preprocessing.text import Tokenizer
from tensorflow.keras.preprocessing.sequence import pad_sequences
```

```python
# Load cleaned dataset
data = pd.read_csv("../data/mobixa_intents_clean.csv")



data.head()
```

```python
X = data["processed_text"]

print(X.head())
```

```python
bow = CountVectorizer()

X_bow = bow.fit_transform(X)
```

```python
print("Bag of Words Shape")

print(X_bow.shape)
```

```python
print(bow.get_feature_names_out())
```

```python
bow_df = pd.DataFrame(
    X_bow.toarray(),
    columns=bow.get_feature_names_out()
)

bow_df.head()
```

```python
with open("../models/bow_vectorizer.pkl","wb") as file:
    pickle.dump(bow,file)

print("Bag of Words saved successfully.")
```

```python
tfidf = TfidfVectorizer()

X_tfidf = tfidf.fit_transform(X)
```

```python
print("TF-IDF Shape")

print(X_tfidf.shape)
```

```python
print(tfidf.get_feature_names_out())
```

```python
tfidf_df = pd.DataFrame(
    X_tfidf.toarray(),
    columns=tfidf.get_feature_names_out()
)

tfidf_df.head()
```

```python
with open("../models/tfidf_vectorizer.pkl","wb") as file:
    pickle.dump(tfidf,file)

print("TF-IDF saved successfully.")
```

```python
tokenizer = Tokenizer()

tokenizer.fit_on_texts(X)
```

```python
print(tokenizer.word_index)
```

```python
sequences = tokenizer.texts_to_sequences(X)

print(sequences[:5])
```

```python
max_length = max(len(sentence) for sentence in sequences)

X_pad = pad_sequences(
    sequences,
    maxlen=max_length,
    padding="post"
)

print(X_pad[:5])
```

```python
print("Word Embedding Shape")

print(X_pad.shape)
```

```python
with open("../models/tokenizer.pkl","wb") as file:
    pickle.dump(tokenizer,file)

print("Tokenizer saved successfully.")
```

```python
print("Feature Engineering Completed Successfully")

print("--------------------------------")

print("Bag of Words Shape :", X_bow.shape)

print("TF-IDF Shape :", X_tfidf.shape)

print("Word Embedding Shape :", X_pad.shape)
```
